```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dt = pd.read_csv("titanic_dataset.csv")
```

```python
dt
```

```python
dt.info()
```

```python
print("Number of rows and columns:", dt.shape)
```

```python
dt.set_index("PassengerId", inplace=True)
```

```python
dt.describe()
```

```python
print("Survived:")
print(dt["Survived"].value_counts())

print("\nPassenger Class:")
print(dt["Pclass"].value_counts())

print("\nGender:")
print(dt["Sex"].value_counts())

print("\nEmbarked:")
print(dt["Embarked"].value_counts())

print("\nSibSp:")
print(dt["SibSp"].value_counts())

print("\nParch:")
print(dt["Parch"].value_counts())
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=dt, x="Survived")

plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Count")

plt.show()
```

```python
print("Unique values in Pclass:")
print(dt["Pclass"].unique())
```

```python
dt.rename(columns={"Sex": "Gender"}, inplace=True)

dt
```

```python
sns.catplot(
    data=dt,
    x="Pclass",
    y="Survived",
    hue="Gender",
    kind="bar"
)

plt.title("Pclass vs Survived by Gender")
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(8, 5))

graph = sns.countplot(
    data=dt,
    x="Pclass",
    hue="Survived",
    ax=ax1
)

for p in graph.patches:
    height = p.get_height()

    graph.text(
        p.get_x() + p.get_width()/2,
        height + 5,
        int(height),
        ha="center"
    )

plt.title("Passenger Class and Survival Count")
plt.xlabel("Passenger Class")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=dt,
    x="Survived",
    y="Age"
)

plt.title("Age vs Survived")
plt.xlabel("Survived")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=dt,
    x="Pclass",
    y="Age",
    hue="Gender"
)

plt.title("Pclass vs Age by Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Age")

plt.show()
```

```python
sns.catplot(
    data=dt,
    x="Pclass",
    y="Survived",
    hue="Gender",
    kind="bar",
    height=5,
    aspect=1.5
)

plt.title("Pclass vs Survived by Gender")

plt.show()
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    dt.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.pairplot(
    dt,
    vars=["Age", "Fare", "Pclass", "Survived"],
    hue="Gender"
)

plt.show()
```
